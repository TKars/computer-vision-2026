# ConvNeXt Tiny

128×128, один канал, pretrained ImageNet-12k → ImageNet-1k. AdamW 5e-4, weight decay 0.05, batch 256, 20 эпох, OneCycle, drop path 0.3, mixup 0.2, label smoothing 0.1, EMA 0.9995 с warmup.

Train 655500 / val 34500, прежняя разбивка. Все классы с одинаковым весом loss.


```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import display

output = Path("output/convnext-tiny95-20261016")
ensemble = Path("output/convnext-ensemble95-20261016")
config = json.loads((output / "config.json").read_text())
config
```

```python
history = pd.read_csv(output / "history.csv")
display(history)
display(pd.read_csv(output / "scores.csv"))
```

```python
verification = json.loads((output / "verification.json").read_text())
assert verification["val_examples"] == 34500
assert verification["submission_rows"] == 172500
assert verification["submission_differences"] == 0
verification
```

```python
summary = json.loads((ensemble / "summary.json").read_text())
display(pd.DataFrame(summary["baselines"]))
display(pd.DataFrame(summary["best_by_mode"]).T)
summary["selected"]
```

```python
verified = json.loads((ensemble / "verification.json").read_text())
assert verified["accuracy"] == summary["selected"]["accuracy"]
assert verified["val_examples"] == 34500
assert verified["recipe_prediction_differences"] == 0
submission_check = json.loads((ensemble / "submission_verification.json").read_text())
assert submission_check["submission_rows"] == 172500
assert submission_check["submission_differences"] == 0
display(submission_check)
display(pd.read_csv(ensemble / "submission.csv").head())
```
